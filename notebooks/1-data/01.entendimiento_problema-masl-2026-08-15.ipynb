# Entendimiento del Problema y las Fuentes de Datos
## Predicción de Precios de Casas en Boston

Este análisis se basa en el archivo `Precios_Casas_Boston.csv`, correspondiente al conjunto de datos clásico **Boston Housing** (448 registros, 14 atributos + ID), con `medv` (valor mediano de la vivienda en miles de USD) como variable objetivo. El dataset presenta valores faltantes (entre 1 y 16 por columna) y al menos un valor atípico evidente en `crim` (máximo de 15,288, muy por encima del rango esperado).

---

### ¿Cuál es el objetivo del problema?

Predecir el precio (valor mediano, `medv`) de una vivienda en los suburbios de Boston a partir de sus atributos estructurales, socioeconómicos y ambientales: criminalidad, zonificación, uso industrial, cercanía al río Charles, contaminación (`nox`), número de cuartos (`rm`), edad de la vivienda, distancia a centros de empleo, accesibilidad a autopistas, impuestos, ratio alumno-profesor, composición racial del vecindario y % de población de estatus socioeconómico bajo.

### ¿Cómo se usará su solución?

Típicamente para apoyar decisiones de tasación/valoración inmobiliaria: agentes, compradores o entidades de crédito hipotecario podrían usarla como referencia de precio "justo" antes de negociar, o como insumo dentro de un sistema más amplio (por ejemplo, un modelo de riesgo crediticio hipotecario o un dashboard de mercado inmobiliario).

### ¿Cuáles son las soluciones actuales (si las hay)?

Sin más contexto de negocio, no es posible saber si existe una solución actual (¿tasación manual por peritos? ¿otro modelo ya en uso?). Esto queda como un **supuesto pendiente por confirmar** con el dueño del problema.

### ¿Cómo se debe enmarcar este problema?

- **Aprendizaje supervisado**: existe una variable objetivo etiquetada (`medv`).
- **Tarea de regresión**: la variable a predecir es continua, no categórica.
- **Aprendizaje por lotes (offline)**: no hay indicios de que los datos lleguen en flujo continuo; el modelo se reentrenaría periódicamente, no en tiempo real.

### ¿Cómo se debe medir el desempeño? (primera intuición)

- **RMSE** (raíz del error cuadrático medio): la métrica más común para este dataset, penaliza fuerte los errores grandes.
- **MAE** (error absoluto medio): más interpretable en unidades de dólares.
- **R²**: como complemento, para dimensionar el % de varianza explicada por el modelo.

### ¿La medida de desempeño está alineada con el objetivo del problema?

Parcialmente. RMSE penaliza con más fuerza los errores en casas caras, lo cual puede o no ser deseable según el uso real: si el objetivo es evitar errores grandes en cualquier segmento de precio, MAE o incluso un error porcentual (MAPE) podría alinearse mejor, especialmente dado el rango amplio de precios (5 a 50, en miles de USD).

### ¿Cuál sería el desempeño mínimo necesario para alcanzar el objetivo?

No se puede determinar sin conocer el caso de uso de negocio (¿cuánto error en dólares es tolerable para quien tome la decisión final?). Queda **pendiente de definir** junto con el "cliente" o usuario final del modelo.

### ¿Cuáles son los problemas parecidos? ¿Se pueden reutilizar experiencias o herramientas ya creadas?

Sí. Es un problema clásico de **tasación automatizada de bienes raíces** (AVM, *Automated Valuation Model*), con abundante literatura y modelos base ya probados sobre este mismo dataset (regresión lineal, árboles de decisión, XGBoost, redes neuronales), lo que permite partir de benchmarks conocidos.

### ¿Hay experiencia del problema disponible?

### (Importante) ¿Cómo se puede resolver el problema manualmente?

Un tasador experto combinaría reglas empíricas, por ejemplo:
- Precio base por m² según la zona.
- Ajuste positivo por número de habitaciones/cuartos (`rm`).
- Penalización por antigüedad (`age`) y por criminalidad (`crim`).
- Prima por cercanía a buenos colegios (`ptratio` bajo) y a zonas verdes/río (`chas`).
- Descuento por cercanía a zonas industriales (`indus`) o alta contaminación (`nox`).

En esencia, sería una fórmula de regresión construida "a mano" con base en la experiencia de mercado.
