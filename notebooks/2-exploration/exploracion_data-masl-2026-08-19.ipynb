## Hallazgos iniciales

- El dataset tiene 448 filas y 15 columnas (incluyendo `ID`).
- Todas las columnas se cargaron como `float64` o `int64`; no hay columnas tipo `object` o string.
- Las columnas `crim`, `zn`, `indus`, `chas`, `nox`, `rm`, `age`, `dis`, `rad`, `tax`,
  `medv` presentan valores nulos (entre 1 y 16 registros por columna), representados
  directamente como `NaN` — no hay un valor centinela como `"?"`.
- **Problema de escala detectado**: las columnas `nox` (proporción, rango esperado 0.4–0.9), `rm` (número de cuartos, rango esperado ~3–9), `crim` (proporción de rango esperado ~0.006 – 90 (tasa de criminalidad per cápita))  y `dis` (proporción de rango esperado 	1.13 – 12.13)   tienen valores muy por fuera de rango (Hasta 15,288), consistente con un corrimiento del punto decimal en una parte de los registros. Se corrige en la sección de conversión de tipos.
- `chas` tiene solo 2 valores únicos (0/1) → es en realidad una variable booleana/categórica,
  no continua.
- `rad` tiene 9 valores únicos, incluyendo `24` como posible categoría atípica → revisar
  si es un índice de accesibilidad discreto (categórico ordinal) en vez de numérico continuo.

### Nota histórica sobre la variable `black`

Esta variable es controversial y ampliamente criticada en la literatura de ciencia de datos
moderna. Fue construida por Harrison & Rubinfeld (1978) bajo el supuesto de que la relación
entre la proporción de población afrodescendiente y el valor de la vivienda no es lineal sino
cuadrática, centrada artificialmente en 0.63. Por esta razón, scikit-learn eliminó este dataset
de sus datasets de ejemplo en 2020, citando el sesgo racial codificado directamente en la
construcción de esta variable.

### Supuesto de corrección de escala

Se identificó que un subconjunto de valores en `crim`, `nox`, `rm` y `dis` está fuera del
rango físicamente esperado para cada variable (según las definiciones oficiales del dataset
en `Informacion.txt` y la documentación original de Harrison & Rubinfeld, 1978), de forma
consistente con un corrimiento del punto decimal (× 1000).

**Este comportamiento no está documentado en `Informacion.txt` provisto por el profesor** —
se trata de un hallazgo propio durante la exploración, no de un problema conocido/esperado
del ejercicio. Se asume que estos valores deben dividirse entre 1000 para volver al rango
esperado, dado que las demás columnas (`zn`, `indus`, `age`, `tax`, `ptratio`, `black`,
`lstat`, `medv`) sí caen dentro de sus rangos oficiales sin corrección, lo que sugiere un
error de captura/transformación aislado a estas cuatro columnas específicas y no un problema
generalizado del dataset.

```python
# Comparación de rangos: valores "sanos" vs "atípicos" por columna, para justificar el factor de escala
for col, umbral in [("crim", 100), ("nox", 1), ("rm", 15), ("dis", 20)]:
    sanos = boston_df.loc[boston_df[col] <= umbral, col]
    atipicos = boston_df.loc[boston_df[col] > umbral, col]
    print(
        f"{col}: sanos [{sanos.min():.3f} - {sanos.max():.3f}]  |  atípicos÷1000 [{atipicos.min() / 1000:.3f} - {atipicos.max() / 1000:.3f}]"
    )
```

```python
# Umbrales de escala física esperada por variable (ver hallazgo documentado arriba)
UMBRAL_CRIM = 100  # tasa de criminalidad per cápita
UMBRAL_RM = 15  # cuartos promedio por vivienda
UMBRAL_DIS = 20  # distancia ponderada a centros de empleo
FACTOR_CORRECCION = 1000

boston_df.loc[boston_df["crim"] > UMBRAL_CRIM, "crim"] = boston_df["crim"] / FACTOR_CORRECCION
boston_df.loc[boston_df["rm"] > UMBRAL_RM, "rm"] = boston_df["rm"] / FACTOR_CORRECCION
boston_df.loc[boston_df["dis"] > UMBRAL_DIS, "dis"] = boston_df["dis"] / FACTOR_CORRECCION

boston_df[["crim", "nox", "rm", "dis"]].describe()
```

## Categorización de variables

```python
cols_boolean = ["chas"]
boston_df[cols_boolean] = boston_df[cols_boolean].astype(
    "boolean"
)  # nullable boolean, preserva NaN
```

```python
cols_categoric_ordinal = ["rad"]
boston_df[cols_categoric_ordinal] = boston_df[cols_categoric_ordinal].astype(
    "Int64"
)  # nullable int, preserva NaN
```

```python
cols_numeric_float = [
    "crim",
    "zn",
    "indus",
    "nox",
    "rm",
    "age",
    "dis",
    "tax",
    "ptratio",
    "black",
    "lstat",
    "medv",
]
boston_df[cols_numeric_float] = boston_df[cols_numeric_float].astype("float64")
```
